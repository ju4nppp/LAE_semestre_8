```python
import pandas as pd
```

PARTE 1

```python
df = pd.read_csv("/datos_feo_bien.csv")
df['es_kahlua'] = (df['sabor'] == 'Kahlúa').astype(int)
df['es_horchata'] = (df['sabor'] == 'horchata').astype(int)
df['tamaño'] = df['tamaño'].map({'chica': 0, 'grande': 1})
df['hielo'] = df['hielo'].map({'sin': 0, 'con': 1})
df.head()
```

```python
import statsmodels.api as sm


X = df.drop("score", axis=1)
X = X.drop("sabor", axis=1)


y = df["score"]

# Add a constant to the independent variables (X) for the OLS model
X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()

print(results.summary())
```

```python
df = pd.read_csv("/datos_feo_bien.csv")
df['es_kahlua'] = (df['sabor'] == 'Kahlúa').astype(int)
df['es_horchata'] = (df['sabor'] == 'horchata').astype(int)
df['tamaño'] = df['tamaño'].map({'chica': 1, 'grande': -1})
df['hielo'] = df['hielo'].map({'sin': -1, 'con': 1})
df
```

### Conclusiones

Al cambiar el enconding para las variables de hielo y tamaño de 0 y 1, a -1 y 1, subió el intercepto del modelo por exactamente 1 punto. Esto es debido a que el intercepto representa el valor predicho cuando todas las variables explicativas son 0. En el modelo 1 se calcula el caso base, de bebida sin hielo y chica. En el modelo 2 se calcula el caso promedio por que el cero seria la mitad entre -1 y 1, y como las características extra de hielo y tamaño añaden valor, el promedio sube.
