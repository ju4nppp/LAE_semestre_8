# Optimización Bayesiana

Realizamos una regresión de proceso gaussiano para encontrar distribuciones que podrían darnos estas muestras

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process import kernels
import matplotlib.pyplot as plt
import numpy as np
```

```python
#datos iniciales
X = np.array([[10], [2], [3], [7], [12]]).reshape(-1, 1)
y = np.array([2847, -4017, -1255, 1773, 563])
```

```python
#fitteamos el modelo para los datos actuales
gpr = GaussianProcessRegressor(n_restarts_optimizer=10, random_state=42)

gpr.fit(X, y)
print("GaussianProcessRegressor model fitted successfully with new X and y.")
```

```python
#simulamos puntos de X y fiteamos con el regresor
Xsim = np.linspace(2, 15, 10001).reshape(-1, 1)
ysim, sigma = gpr.predict(Xsim, return_std=True)
```

```python
#creamos y_upper con 2 intervalos de confianza (95%)
y_upper = ysim + 2*sigma
#buscamos el maximo con y_upper
maximo = np.argmax(y_upper)
Xsim[maximo]
```

```python
#repetimos el proceso con la nueva "y", que nos arroja la caja negra con nuestro máximo encontrado
X2 = np.array([[10], [2], [3], [7], [12], [9.9963]]).reshape(-1, 1)
y2 = np.array([2847, -4017, -1255, 1773, 563, 2847.31363])

gpr.fit(X2, y2)

Xsim2 = np.linspace(2, 15, 10001).reshape(-1, 1)
ysim2, sigma2 = gpr.predict(Xsim2, return_std=True)

y_upper2 = ysim2 + 2*sigma2
maximo2 = np.argmax(y_upper2)
Xsim2[maximo2]
```
