## 1.Tomar advertising y hacer una regresión lineal simple

```python
df_adv = pd.read_csv("/content/Advertising.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df_adv.head()
```

```python
LR = LinearRegression()
```

```python
X, y = df_adv[["TV", "radio", "newspaper"]], df_adv["sales"]
```

```python
LR.fit(X, y)
y_pred = LR.predict(X)
```

## 2. Calcular R2 de (1)

```python
r2_LR = r2_score(y, y_pred)
print(r2_LR)
```

## 3. Usar 1 árbol de decisión para regresión sobre el mismo dataset

```python
Tree = DecisionTreeRegressor()
Tree.fit(X, y)
y_pred_tree = Tree.predict(X)
```

## 4. Calcular R2 de (3)


```python
r2_tree = r2_score(y, y_pred_tree)
print(r2_tree)
```

## Repetir (1-4) con train_test_split 50/50

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
# Con LR
LR.fit(X_train, y_train)
y_pred_train = LR.predict(X_train)
y_pred_test = LR.predict(X_test)

r2_LR_train = r2_score(y_train, y_pred_train)
r2_LR_test = r2_score(y_test, y_pred_test)

print("R2 en entrenamiento:", r2_LR_train)
print("R2 en prueba:", r2_LR_test)
```

```python
# Con Tree
Tree.fit(X_train, y_train)
y_pred_train_tree = Tree.predict(X_train)
y_pred_test_tree = Tree.predict(X_test)

r2_LR_train_tree = r2_score(y_train, y_pred_train_tree)
r2_LR_test_tree = r2_score(y_test, y_pred_test_tree)

print("R2 en entrenamiento:", r2_LR_train_tree)
print("R2 en prueba:", r2_LR_test_tree)
```
