## Ejercicio 1

 Using basic statistical properties of the variance, as well as singlevariable calculus, derive (5.6). In other words, prove that α given by
(5.6) does indeed minimize Var(αX + (1 − α)Y ).

**Objetivo:** Demostrar que $\alpha = \frac{\sigma_Y^2 - \sigma_{XY}}{\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}}$ minimiza la varianza de la expresión $\text{Var}(\alpha X + (1 - \alpha)Y)$.

**Paso 1: Expandir la expresión de la varianza.**
Utilizando las propiedades básicas de la varianza, sabemos que para dos variables aleatorias $A$ y $B$, y constantes $a$ y $b$:
$$ \text{Var}(aA + bB) = a^2\text{Var}(A) + b^2\text{Var}(B) + 2ab\text{Cov}(A, B) $$

Sustituyendo $A = X$, $B = Y$, $a = \alpha$ y $b = (1 - \alpha)$, definimos una función $f(\alpha)$ que representa la varianza que queremos minimizar:
$$ f(\alpha) = \text{Var}(\alpha X + (1 - \alpha)Y) $$
$$ f(\alpha) = \alpha^2 \text{Var}(X) + (1 - \alpha)^2 \text{Var}(Y) + 2\alpha(1 - \alpha)\text{Cov}(X,Y) $$

Sustituyendo con la notación del libro ($\sigma_X^2 = \text{Var}(X)$, $\sigma_Y^2 = \text{Var}(Y)$, y $\sigma_{XY} = \text{Cov}(X,Y)$):
$$ f(\alpha) = \alpha^2 \sigma_X^2 + (1 - \alpha)^2 \sigma_Y^2 + 2\alpha(1 - \alpha)\sigma_{XY} $$

**Paso 2: Derivar la función con respecto a $\alpha$.**
Para encontrar el valor que minimiza la función, aplicamos las reglas de cálculo de una variable y derivamos $f(\alpha)$ con respecto a $\alpha$:
$$ \frac{d}{d\alpha} f(\alpha) = \frac{d}{d\alpha} \left[ \alpha^2 \sigma_X^2 + (1 - 2\alpha + \alpha^2) \sigma_Y^2 + (2\alpha - 2\alpha^2)\sigma_{XY} \right] $$
$$ \frac{d}{d\alpha} f(\alpha) = 2\alpha \sigma_X^2 + 2(1 - \alpha)(-1)\sigma_Y^2 + (2 - 4\alpha)\sigma_{XY} $$
$$ \frac{d}{d\alpha} f(\alpha) = 2\alpha \sigma_X^2 - 2(1 - \alpha)\sigma_Y^2 + 2(1 - 2\alpha)\sigma_{XY} $$

**Paso 3: Igualar la derivada a cero para encontrar el punto crítico.**
$$ 2\alpha \sigma_X^2 - 2(1 - \alpha)\sigma_Y^2 + 2(1 - 2\alpha)\sigma_{XY} = 0 $$

Dividimos toda la ecuación entre $2$ para simplificarla:
$$ \alpha \sigma_X^2 - (1 - \alpha)\sigma_Y^2 + (1 - 2\alpha)\sigma_{XY} = 0 $$
$$ \alpha \sigma_X^2 - \sigma_Y^2 + \alpha \sigma_Y^2 + \sigma_{XY} - 2\alpha\sigma_{XY} = 0 $$

**Paso 4: Despejar $\alpha$.**
Agrupamos todos los términos que contienen $\alpha$ del lado izquierdo de la ecuación y pasamos el resto al lado derecho:
$$ \alpha \sigma_X^2 + \alpha \sigma_Y^2 - 2\alpha\sigma_{XY} = \sigma_Y^2 - \sigma_{XY} $$

Factorizamos $\alpha$:
$$ \alpha (\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}) = \sigma_Y^2 - \sigma_{XY} $$

Finalmente, dividimos para aislar $\alpha$:
$$ \alpha = \frac{\sigma_Y^2 - \sigma_{XY}}{\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}} $$

**Paso 5: Comprobar que el punto crítico es un mínimo (Prueba de la segunda derivada).**
Para asegurar que este valor realmente minimiza la varianza (y no la maximiza), calculamos la segunda derivada de $f(\alpha)$:
$$ f''(\alpha) = \frac{d}{d\alpha} \left[ 2\alpha \sigma_X^2 - 2\sigma_Y^2 + 2\alpha\sigma_Y^2 + 2\sigma_{XY} - 4\alpha\sigma_{XY} \right] $$
$$ f''(\alpha) = 2\sigma_X^2 + 2\sigma_Y^2 - 4\sigma_{XY} $$
$$ f''(\alpha) = 2(\sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY}) $$

Por definición, sabemos que la varianza de cualquier variable aleatoria es siempre mayor o igual a cero. Si tomamos la varianza de la diferencia entre $X$ e $Y$, su expansión es:
$$ \text{Var}(X - Y) = \text{Var}(X) + \text{Var}(Y) - 2\text{Cov}(X,Y) = \sigma_X^2 + \sigma_Y^2 - 2\sigma_{XY} $$
Como $\text{Var}(X - Y) \ge 0$, entonces la segunda derivada $f''(\alpha)$ es positiva (asumiendo que $X$ e $Y$ no están perfectamente correlacionadas con la misma varianza). Al ser la segunda derivada positiva, la función es estrictamente convexa, lo que confirma que el valor obtenido para $\alpha$ representa el mínimo global de la varianza. Q.E.D.

We will now derive the probability that a given observation is part
of a bootstrap sample. Suppose that we obtain a bootstrap sample
from a set of n observations.



### (a) What is the probability that the first bootstrap observation is not the jth observation from the original sample? Justify your answer.

**Objetivo:** Determinar la probabilidad de que la primera observación de la muestra bootstrap no sea la $j$-ésima observación de la muestra original.

**Justificación paso a paso:**

En el proceso de creación de una muestra bootstrap, las observaciones se extraen de la muestra original (que tiene un tamaño de $n$ elementos) de forma aleatoria y **con reemplazo**. Esto significa que en cualquier extracción individual, cada una de las $n$ observaciones de la muestra original tiene exactamente la misma probabilidad de ser seleccionada.

La probabilidad de que la primera observación elegida sea exactamente la $j$-ésima observación es:
$$ P(\text{seleccionar la } j\text{-ésima observación}) = \frac{1}{n} $$

Dado que estamos buscando la probabilidad de que esto **no** ocurra, debemos calcular el evento complementario. La suma de las probabilidades de todos los resultados posibles es $1$, por lo que restamos la probabilidad de seleccionarla de $1$:
$$ P(\text{no seleccionar la } j\text{-ésima observación}) = 1 - \frac{1}{n} $$

**Respuesta final:**
La probabilidad de que la primera observación bootstrap no sea la $j$-ésima observación de la muestra original es $1 - \frac{1}{n}$.

### (b) What is the probability that the second bootstrap observation is not the jth observation from the original sample?


**Objetivo:** Determinar la probabilidad de que la segunda observación de la muestra bootstrap no sea la $j$-ésima observación de la muestra original.

**Justificación paso a paso:**

La característica fundamental del muestreo bootstrap es que las extracciones se realizan **con reemplazo** y son eventos **independientes**. Esto significa que, después de extraer la primera observación, esta "regresa" al conjunto original, por lo que el conjunto de donde se toma la segunda observación vuelve a tener exactamente los mismos $n$ elementos iniciales.

El resultado de la primera extracción no afecta de ninguna manera el resultado de la segunda. Por lo tanto, las probabilidades se mantienen idénticas a las del inciso anterior.

La probabilidad de seleccionar la $j$-ésima observación en el segundo intento es:
$$ P(\text{seleccionar la } j\text{-ésima observación en el segundo intento}) = \frac{1}{n} $$

Aplicando la regla del complemento para encontrar la probabilidad de que este evento no ocurra:
$$ P(\text{no seleccionar la } j\text{-ésima observación en el segundo intento}) = 1 - \frac{1}{n} $$

**Respuesta final:**
La probabilidad de que la segunda observación bootstrap no sea la $j$-ésima observación de la muestra original es $1 - \frac{1}{n}$.

### (c) Argue that the probability that the jth observation is not in the bootstrap sample is (1 − 1/n)**n.


**Objetivo:** Argumentar que la probabilidad de que la $j$-ésima observación no se encuentre en toda la muestra bootstrap es $(1 - 1/n)^n$.

**Justificación paso a paso:**

Como hemos establecido en los incisos (a) y (b), la probabilidad de que la $j$-ésima observación no sea seleccionada en una sola extracción cualquiera es constante y equivale a $1 - \frac{1}{n}$.

Una muestra bootstrap completa requiere que este proceso de extracción se repita $n$ veces en total. Dado que el muestreo se realiza con reemplazo, cada una de estas $n$ extracciones es un evento estadísticamente **independiente** de los demás.

Por la regla de la multiplicación para eventos independientes, la probabilidad de que ocurran múltiples eventos independientes en secuencia es igual al producto de sus probabilidades individuales.

Por lo tanto, la probabilidad de que la $j$-ésima observación no sea elegida en la primera extracción, **Y** no sea elegida en la segunda, **Y** no sea elegida en la tercera, y así sucesivamente hasta la $n$-ésima extracción, se calcula multiplicando la probabilidad de no ser seleccionada $n$ veces:

$$ P(\text{no en la muestra bootstrap}) = P(\text{no en la 1ra}) \times P(\text{no en la 2da}) \times \dots \times P(\text{no en la } n\text{-ésima}) $$

Sustituyendo los valores:

$$ P(\text{no en la muestra bootstrap}) = \left(1 - \frac{1}{n}\right) \times \left(1 - \frac{1}{n}\right) \times \dots \times \left(1 - \frac{1}{n}\right) \quad \text{($n$ veces)} $$

Lo que se simplifica algebraicamente como:

$$ P(\text{no en la muestra bootstrap}) = \left(1 - \frac{1}{n}\right)^n $$

Con esto queda argumentado y demostrado.

### (d) When n = 5, what is the probability that the jth observation is in the bootstrap sample?

**Objetivo:** Calcular la probabilidad de que la $j$-ésima observación **sí** se encuentre en la muestra bootstrap cuando el tamaño de la muestra es $n = 5$.

**Justificación paso a paso:**

Basándonos en el resultado del inciso (c), sabemos que la probabilidad de que la $j$-ésima observación **no** esté en la muestra bootstrap (después de $n$ extracciones) es de $(1 - 1/n)^n$.

El evento de que la observación **sí** esté en la muestra (al menos una vez) es el complemento exacto de que no esté en absoluto. Por lo tanto, la probabilidad de que esté presente se calcula restando la probabilidad del inciso anterior a $1$:

$$ P(\text{en la muestra}) = 1 - P(\text{no en la muestra}) $$
$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{n}\right)^n $$

Ahora, sustituimos el valor dado de $n = 5$ en nuestra fórmula:

$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{5}\right)^5 $$
$$ P(\text{en la muestra}) = 1 - \left(\frac{4}{5}\right)^5 $$
$$ P(\text{en la muestra}) = 1 - (0.8)^5 $$

Calculamos el valor de la potencia y la resta final:

$$ P(\text{en la muestra}) = 1 - 0.32768 $$
$$ P(\text{en la muestra}) = 0.67232 $$

**Respuesta final:**
Cuando $n = 5$, la probabilidad de que la $j$-ésima observación esté en la muestra bootstrap es de **$0.67232$** (o un $67.23\%$).

### (e) When n = 100, what is the probability that the jth observation is in the bootstrap sample?

**Objetivo:** Calcular la probabilidad de que la $j$-ésima observación se encuentre en la muestra bootstrap cuando el tamaño de la muestra es $n = 100$.

**Justificación paso a paso:**

Utilizando la misma lógica y la fórmula derivada en los incisos anteriores, sabemos que la probabilidad de que una observación específica sea seleccionada al menos una vez es el complemento de la probabilidad de que no sea seleccionada en ninguna de las $n$ extracciones.

La fórmula general es:
$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{n}\right)^n $$

Ahora, sustituimos el nuevo tamaño de muestra $n = 100$:
$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{100}\right)^{100} $$

Calculamos el valor numérico de la potencia =
$$ (0.99)^{100} \approx 0.36603 $$

Finalmente, realizamos la resta:
$$ P(\text{en la muestra}) \approx 1 - 0.36603 $$
$$ P(\text{en la muestra}) \approx 0.63397 $$

**Respuesta final:**
Cuando $n = 100$, la probabilidad de que la $j$-ésima observación esté en la muestra bootstrap es de aproximadamente **0.63397** (o un 63.4%).

### (f) When n = 10, 000, what is the probability that the jth observation is in the bootstrap sample?

**Objetivo:** Calcular la probabilidad de que la $j$-ésima observación se encuentre en la muestra bootstrap cuando el tamaño de la muestra es $n = 10,000$.

**Justificación paso a paso:**

Nuevamente, aplicamos la fórmula general que derivamos en el inciso (c) y que usamos en (d) y (e):
$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{n}\right)^n $$

Sustituimos el valor de $n = 10,000$:
$$ P(\text{en la muestra}) = 1 - \left(1 - \frac{1}{10000}\right)^{10000} $$
$$ P(\text{en la muestra}) = 1 - (0.9999)^{10000} $$

Calculamos el valor de la potencia:
$$ (0.9999)^{10000} \approx 0.36786 $$

Realizamos la resta final:
$$ P(\text{en la muestra}) \approx 1 - 0.36786 $$

*Nota adicional:* A medida que $n$ crece hacia el infinito, la expresión $\left(1 - \frac{1}{n}\right)^n$ converge matemáticamente a $\frac{1}{e}$ (donde $e$ es el número de Euler, aproximadamente $2.71828$). Por lo tanto, $1 - \frac{1}{e} \approx 1 - 0.367879 = 0.63212$. El resultado con $n=10,000$ ya es extremadamente cercano a este límite asintótico, conocido familiarmente como la "regla del 63.2%" en el muestreo bootstrap.

### (h) We will now investigate numerically the probability that a bootstrap sample of size n = 100 contains the jth observation. Here j = 4. We first create an array `store` with values that will subsequently be overwritten using the function `np.empty()`. We then repeatedly create bootstrap samples, and each time we record whether or not the fifth observation is contained in the bootstrap sample. Comment on the results obtained.

**Objetivo:** Investigar numéricamente la probabilidad de que una muestra bootstrap de tamaño $n = 100$ contenga la $j$-ésima observación (en este caso $j = 4$) mediante la repetición del experimento 10,000 veces, y comentar el resultado.

```python
import numpy as np

# Inicializar el generador de números aleatorios
rng = np.random.default_rng(10)

# Crear el arreglo vacío para guardar los resultados
store = np.empty(10000)

for i in range(10000):
    # CORRECCIÓN: Se añade size=100 para extraer una muestra de tamaño n=100
    store[i] = np.sum(rng.choice(100, size=100, replace=True) == 4) > 0

# Calcular y mostrar la media
prob_empirica = np.mean(store)
print(f"Probabilidad numérica obtenida: {prob_empirica}")
```



**Comentarios sobre los resultados obtenidos:**

Podemos observar y concluir lo siguiente:

1. **Confirmación empírica:** El resultado numérico concuerda casi a la perfección con nuestra deducción matemática del inciso (e). En dicho inciso, calculamos teóricamente que para $n = 100$, la probabilidad exacta era $1 - (1 - 1/100)^{100} \approx 0.634$.
2. **Convergencia del Bootstrap:** Esta simulación comprueba que el método de remuestreo funciona tal cual predice la teoría de la probabilidad. Incluso con variaciones aleatorias (las 10,000 iteraciones), la proporción de muestras que contienen la observación de interés gravita hacia la constante matemática del $63.2\%$ ($1 - 1/e$), que es la huella digital del proceso Bootstrap.

## Ejercico 3.
We now review k-fold cross-validation.


### (a) Explain how k-fold cross-validation is implemented.

División aleatoria: El conjunto de datos original de $n$ observaciones se divide aleatoriamente en $k$ grupos (o folds) de tamaño aproximadamente igual, que no se superponen.

Proceso iterativo: Se realiza un bucle que se repite $k$ veces. En cada iteración $i$:

*   El grupo $i$ se reserva y se utiliza como conjunto de validación (testing set).
*   Los $k-1$ grupos restantes se combinan para formar el conjunto de entrenamiento (training set).


*   El modelo estadístico se ajusta (se entrena) utilizando únicamente el conjunto de entrenamiento.
*   Se utiliza el modelo entrenado para predecir las observaciones en el grupo $i$ reservado.

*   Se calcula una métrica de error para esta iteración, típicamente el Error Cuadrático Medio ($\text{MSE}_i$)

Cálculo del error final: Una vez que los $k$ grupos han sido utilizados exactamente una vez como conjunto de validación, se obtiene el error estimado de validación cruzada promediando los $k$ errores individuales calculados:$$ \text{CV}_{(k)} = \frac{1}{k} \sum_{i=1}^k \text{MSE}_i $$






### (b) What are the advantages and disadvantages of k-fold crossvalidation relative to:

i. The validation set approach?


*   **Ventajas**
*   Menor sesgo (Bias): k-fold utiliza una proporción mucho mayor de los datos para entrenar el modelo (por ejemplo, el $90\%$ si $k=10$), en lugar del $50\%$. Esto produce estimaciones menos pesimistas del error de prueba.

*   Menor varianza: En el enfoque simple, la estimación del error depende enormemente de qué observaciones cayeron aleatoriamente en entrenamiento y cuáles en prueba. k-fold estabiliza esta métrica al promediar el error sobre múltiples divisiones diferentes que cubren la totalidad del dataset.
*   Uso eficiente de datos: Cada observación en el dataset original se utiliza tanto para entrenar como para evaluar el modelo en algún punto del proceso.

*   **Desventajas**
*   Costo computacional: k-fold requiere entrenar y evaluar el modelo $k$ veces (por ejemplo, 5 o 10 veces), mientras que el enfoque simple requiere hacerlo solo una vez.



ii. LOOCV?

LOOCV es un caso extremo de k-fold donde el número de particiones es igual al número total de observaciones ($k = n$).

*   **Ventajas de k-fold sobre LOOCV:**

*   Mucho menor costo computacional: LOOCV requiere ajustar el modelo $n$ veces. Si tienes 100,000 observaciones, entrenar el modelo 100,000 veces puede ser prohibitivo. k-fold típicamente usa $k=5$ o $k=10$, reduciendo drásticamente el tiempo de ejecución.
*   Menor varianza en la estimación del error (Trade-off Sesgo-Varianza): En LOOCV, los $n$ conjuntos de entrenamiento son casi idénticos entre sí. El promedio de variables altamente correlacionadas tiene una varianza mayor que el promedio de variables menos correlacionadas. Por ello, k-fold con $k=5$ o $k=10$ a menudo da una estimación más confiable del verdadero error de prueba generalizado que LOOCV.
*   **Desventajas de k-fold sobre LOOCV:**


*   Mayor sesgo: LOOCV entrena con $n-1$ observaciones (casi todo el dataset), por lo que tiene un sesgo prácticamente nulo al estimar el verdadero error. k-fold (ej. $k=5$) entrena con el $80\%$ de los datos, lo que introduce un sesgo ligeramente más alto.


*   No determinista: Dado que k-fold involucra una división inicial aleatoria en $k$ grupos, diferentes ejecuciones darán resultados ligeramente diferentes. LOOCV siempre produce el mismo resultado exacto (es determinista) porque no hay aleatoriedad en cómo se seleccionan los grupos.



## Ejercicio 4.

Suppose that we use some statistical learning method to make a prediction for the response Y for a particular value of the predictor X. Carefully describe how we might estimate the standard deviation of our prediction.


**Objetivo:** Describir detalladamente cómo podemos estimar la desviación estándar de una predicción para la respuesta $Y$ dado un valor particular del predictor $X$.

**Respuesta y justificación:**

Para estimar la desviación estándar de nuestra predicción (también conocida como el error estándar de la predicción), el enfoque más general y robusto aplicable a casi cualquier método de aprendizaje estadístico es utilizar el **método Bootstrap**.

A continuación se describe el procedimiento paso a paso:

1. **Definir el conjunto original:** Partimos de nuestro conjunto de datos original de entrenamiento, el cual contiene $n$ observaciones.
2. **Generar muestras Bootstrap:** Creamos un número grande $B$ de muestras bootstrap (por ejemplo, $B = 1000$). Cada muestra bootstrap se genera extrayendo $n$ observaciones del conjunto de datos original de forma aleatoria y **con reemplazo**.
3. **Ajustar el modelo iterativamente:** Para cada una de las $B$ muestras bootstrap, entrenamos (ajustamos) nuestro método de aprendizaje estadístico desde cero utilizando únicamente los datos de esa muestra en particular.
4. **Generar predicciones:** Una vez entrenado el modelo en la muestra bootstrap $b$ (donde $b = 1, 2, \dots, B$), lo utilizamos para predecir el valor de la respuesta $Y$ para el valor particular de interés de $X$. Denotamos esta predicción individual como $\hat{Y}^{*b}$.
5. **Calcular la desviación estándar:** Al finalizar el proceso, tendremos $B$ predicciones diferentes para el mismo valor de $X$. Estimamos la desviación estándar de nuestra predicción calculando la desviación estándar muestral de estos $B$ valores predichos utilizando la siguiente fórmula:

$$ \text{SE}(\hat{Y}) \approx \sqrt{\frac{1}{B-1} \sum_{b=1}^B \left(\hat{Y}^{*b} - \bar{\hat{Y}}^*\right)^2} $$

Donde $\bar{\hat{Y}}^*$ es la media de todas las $B$ predicciones obtenidas mediante el bootstrap:
$$ \bar{\hat{Y}}^* = \frac{1}{B}\sum_{b=1}^B \hat{Y}^{*b} $$

**Conclusión:** Este método no paramétrico nos permite cuantificar la incertidumbre de nuestra predicción sin necesidad de asumir que los errores siguen una distribución teórica específica (como la distribución normal) o de derivar fórmulas analíticas complejas que podrían no existir para métodos de aprendizaje no lineales.
