```python
import random
import numpy as np
rng = np.random.default_rng()
juegos_7 = 0

def serie():
  wins_a = 0
  wins_b = 0
  games = 0
  while (wins_a < 4) and (wins_b < 4) :
    winner = rng.integers(low=1, high=3)
    games += 1
    if winner == 1:
      wins_a += 1
    else:
      wins_b += 1

  if games >= 7:
    return 1
  else:
    return 0

for i in range(1000000):
  juegos_7 += serie()

proporcion = juegos_7/1000000
print(proporcion)


```

```python
res = 0
positivos = 0
enfermos_positivos = 0


def prueba():
  enfermo = False
  prueba_positiva = False
  if rng.integers(1, 1001) == 1:
    enfermo = True
  ## SI ESTA ENFERMO
  if enfermo:
    if rng.integers(1, 101) in range(1, 99):
      prueba_positiva = True
  ## NO ESTA ENFERMO
  else:
    if rng.integers(1, 101) == 1:
      prueba_positiva = True
    else:
      prueba_positiva = False
  return prueba_positiva, enfermo


for i in range(1000000):
  prueba_positiva, enfermo = prueba()
  if prueba_positiva:
    positivos += 1
    if enfermo:
      enfermos_positivos += 1

print(enfermos_positivos/positivos)


```

```python
def simular_dia():
  realidad_lluvia = rng.random() < 0.25
  testimonios = []
  for i in range(3):
    miente = rng.random() < (1/3)
    if miente:
      testimonios.append(not realidad_lluvia)
    else:
      testimonios.append(realidad_lluvia)

  return realidad_lluvia, all(testimonios)

casos_todos_si = 0
todos_si_y_llovia = 0

for i in range(1000000):
  llovia, todos_si = simular_dia()
  if todos_si:
    casos_todos_si += 1
    if llovia:
      todos_si_y_llovia += 1
res = todos_si_y_llovia/casos_todos_si
print(res)


```
